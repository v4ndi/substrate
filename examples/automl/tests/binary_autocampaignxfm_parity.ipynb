# Binary parity: fmlib и autocampaignxfm

Запускаются полные trainer/evaluator autocampaignxfm и BinaryTask для product и channel. В каждой Optuna study выполняется три trial с seed 42. Сравнение включает ROC AUC, распределение score, найденные гиперпараметры и время. Разные best params допустимы; основной критерий — близость качества и распределения score.

```python
import json
from pathlib import Path

import polars as pl
from IPython.display import display

from tools.automl_parity.binary import run_comparison

FMLIB_ROOT = next(root for root in (Path.cwd(), *Path.cwd().parents) if (root / 'fmlib').is_dir())
WORKSPACE = FMLIB_ROOT.parent
CONFIG_DIR = FMLIB_ROOT / 'examples/automl/tests/configs'
AUTOCAMPAIGN_ROOT = WORKSPACE / 'autocampaignxfm'
AUTOCAMPAIGN_PYTHON = AUTOCAMPAIGN_ROOT / 'env/bin/python'
assert AUTOCAMPAIGN_PYTHON.is_file()
```

```python
comparisons = {}
for scope in ('product', 'channel'):
    comparisons[scope] = run_comparison(
        CONFIG_DIR / f'parity_{scope}_autocampaignxfm.yaml',
        CONFIG_DIR / f'parity_{scope}_fmlib.yaml',
        autocampaign_root=AUTOCAMPAIGN_ROOT,
        autocampaign_python=AUTOCAMPAIGN_PYTHON,
    )
```

```python
metric_rows = []
for scope, comparison in comparisons.items():
    raw_metrics = comparison['metrics']['test_roc_auc']
    calibrated_metrics = comparison['metrics'].get('calibrated_test_roc_auc', {})
    for model_name, values in raw_metrics.items():
        calibrated = calibrated_metrics.get(model_name, {})
        metric_rows.append({
            'scope': scope,
            'model': model_name,
            'autocampaignxfm_roc_auc': values['autocampaignxfm'],
            'fmlib_roc_auc': values['fmlib'],
            'absolute_difference': values['absolute_difference'],
            'calibrated_autocampaignxfm_roc_auc': calibrated.get('autocampaignxfm'),
            'calibrated_fmlib_roc_auc': calibrated.get('fmlib'),
            'calibrated_absolute_difference': calibrated.get('absolute_difference'),
            'score_ks': comparison['score_distribution']['ks_statistic'],
            'calibrated_score_ks': comparison['calibrated_score_distribution']['ks_statistic'],
            'best_params_equal': comparison['best_params_equal'],
            'comparison_passed': comparison['passed'],
        })
display(pl.DataFrame(metric_rows))

timing_rows = []
for scope, comparison in comparisons.items():
    timing_rows.append({
        'scope': scope,
        'autocampaignxfm_train_seconds': comparison['timing_seconds']['autocampaignxfm']['total_train_before_predict'],
        'fmlib_train_seconds': comparison['timing_seconds']['fmlib']['total_train_before_predict'],
        'fmlib_predict_seconds': comparison['timing_seconds']['fmlib']['predict'],
    })
display(pl.DataFrame(timing_rows))
```

Полные отчёты каждого запуска сохранены в `outputs/notebook_tests/parity_product/comparison.json` и `outputs/notebook_tests/parity_channel/comparison.json`. Notebook оставляет видимыми как итоговый флаг, так и отдельные численные расхождения.

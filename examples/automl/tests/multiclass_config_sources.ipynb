# Multiclass configuration sources
Python/YAML configurations with inline and absolute YAML feature lists are normalized and execute one Optuna trial.

```python
import os
from pathlib import Path
from avatar.automl import MulticlassTask, MulticlassTaskConfig
ROOT=Path.cwd().parent.resolve(); TEST=Path.cwd()/'examples/automl/tests'; os.environ['SBER_INFRA_ROOT']=str(ROOT)
data=ROOT/'data/multiclass'; cat=[f'cat_feature_{i}' for i in range(1,6)]; num=[f'num_feature_{i}' for i in range(1,6)]
cat_yaml=(TEST/'configs/features/categorical_columns.yaml').resolve(); num_yaml=(TEST/'configs/features/numerical_columns.yaml').resolve()
space={'iterations':[16],'depth':[2],'learning_rate':[0.1]}
def config(categorical_columns,numerical_columns,name):
 return MulticlassTaskConfig(target_column='target', client_id_column='epk_id', group_column='group', treatment_column='treatment', inverse_treatment=True, report_month_column='report_month', model_scope='product', environment={}, env_type='local',backend='boosting',engine='catboost',device='gpu',categorical_columns=categorical_columns,numerical_columns=numerical_columns,hidden_state_columns=['seq_hidden_state'],hyperopt=True,n_trials=1,optimization_metric='roc_auc_ovr_macro',search_space=space,verbose=False,output_dir=ROOT/f'outputs/notebook_tests/multiclass/config_sources/{name}')
configs=[config(cat,num,'args_inline'),config(cat_yaml,num_yaml,'args_yaml'),MulticlassTaskConfig.from_yaml(TEST/'configs/multiclass_one_trial.yaml'),MulticlassTaskConfig.from_yaml(TEST/'configs/multiclass_inline_features.yaml')]
results=[]
for item in configs:
 task=MulticlassTask(item); training=task.train(data/'train',data/'valid'); prediction=task.predict(data/'test'); evaluation=task.evaluate(data/'test',prediction); artifact=task.save(); results.append((len(item.categorical_columns),len(item.numerical_columns),training.validation_metrics['product'],evaluation.metrics['accuracy'],artifact.exists()))
results
```

# Multiclass CatBoost and XGBoost
Both probability engines execute one Optuna trial, save artifacts, and produce the full multiclass metric set.

```python
from pathlib import Path
import numpy as np
from avatar.automl import MulticlassTask, MulticlassTaskConfig
ROOT=Path.cwd().parent.resolve(); data=ROOT/'data/multiclass'; cat=[f'cat_feature_{i}' for i in range(1,6)]; num=[f'num_feature_{i}' for i in range(1,6)]
spaces={'catboost':{'iterations':[16],'depth':[2]},'xgboost':{'n_estimators':[16],'max_depth':[2]}}
rows=[]
for engine in ('catboost','xgboost'):
 c=MulticlassTaskConfig(target_column='target', client_id_column='epk_id', group_column='group', treatment_column='treatment', inverse_treatment=True, report_month_column='report_month', model_scope='product', environment={}, env_type='local',backend='boosting',engine=engine,device='gpu',categorical_columns=cat,numerical_columns=num,hidden_state_columns=['seq_hidden_state'],hyperopt=True,n_trials=1,optimization_metric='roc_auc_ovr_macro',search_space=spaces[engine],verbose=False,output_dir=ROOT/f'outputs/notebook_tests/multiclass/engines/{engine}')
 t=MulticlassTask(c); training=t.train(data/'train',data/'valid'); prediction=t.predict(data/'test'); evaluation=t.evaluate(data/'test',prediction); artifact=t.save(); score_cols=[f'score_{i}' for i in range(len(prediction.class_order))]; assert np.allclose(prediction.scores.select(score_cols).sum_horizontal(),1.0); rows.append((engine,training.validation_metrics,evaluation.metrics,prediction.class_order,artifact.exists()))
rows
```

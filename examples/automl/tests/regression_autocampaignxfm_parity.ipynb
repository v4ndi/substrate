# Regression terminal parity
The configured reference trainer/evaluator and fmlib workflow run three seeded trials for product and channel. The report compares MSE/MAE/MAPE, score distributions, best parameters, and elapsed time. Native categorical handling may cause small expected differences and is documented in the report.

```python
from pathlib import Path

import polars as pl
from IPython.display import display

from tools.automl_parity.regression import run_comparison

FMLIB_ROOT = next(root for root in (Path.cwd(), *Path.cwd().parents) if (root / 'fmlib').is_dir())
ROOT = FMLIB_ROOT.parent
CONFIGS = FMLIB_ROOT / 'examples/automl/tests/configs'
REFERENCE = ROOT / 'autocampaignxfm'
REFERENCE_PYTHON = REFERENCE / 'env/bin/python'
reports = {}
for scope in ('product', 'channel'):
    reports[scope] = run_comparison(
        CONFIGS / f'parity_regression_{scope}_autocampaignxfm.yaml',
        CONFIGS / f'parity_regression_{scope}_fmlib.yaml',
        autocampaign_root=REFERENCE,
        autocampaign_python=REFERENCE_PYTHON,
    )

metric_rows = []
for scope, report in reports.items():
    for calibrated, key in ((False, 'metrics'), (True, 'calibrated_metrics')):
        metric_rows.append({
            'scope': scope,
            'calibrated': calibrated,
            **{f'{name}_relative_difference': values['relative_difference'] for name, values in report[key].items()},
            'score_ks': report['calibrated_score_distribution' if calibrated else 'score_distribution']['ks_statistic'],
            'best_params_equal': report['best_params_equal'],
            'comparison_passed': report['passed'],
        })
display(pl.DataFrame(metric_rows))
display({scope: report['timing_seconds'] for scope, report in reports.items()})
```

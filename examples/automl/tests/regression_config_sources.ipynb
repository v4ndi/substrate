# Regression configuration sources
Python/YAML and inline/absolute YAML feature lists are normalized and each workflow runs one Optuna trial.

```python
import os
from pathlib import Path
from avatar.automl import RegressionTask, RegressionTaskConfig
ROOT=Path.cwd().parent.resolve(); TEST=Path.cwd()/'examples/automl/tests'; os.environ['SBER_INFRA_ROOT']=str(ROOT)
cat=[f'cat_feature_{i}' for i in range(1,6)]; num=[f'num_feature_{i}' for i in range(1,6)]
cat_yaml=(TEST/'configs/features/categorical_columns.yaml').resolve(); num_yaml=(TEST/'configs/features/numerical_columns.yaml').resolve()
space={'iterations':[16],'depth':[2]}
def python_config(categorical_columns,numerical_columns,name):
 return RegressionTaskConfig(target_column='target', client_id_column='epk_id', group_column='group', treatment_column='treatment', inverse_treatment=True, report_month_column='report_month', model_scope='product', environment={}, env_type='local',backend='boosting',engine='catboost',device='gpu',categorical_columns=categorical_columns,numerical_columns=numerical_columns,hidden_state_columns=['seq_hidden_state'],hyperopt=True,n_trials=1,optimization_metric='mse',search_space=space,verbose=False,output_dir=ROOT/f'outputs/notebook_tests/regression/{name}')
configs={'arguments + inline features':python_config(cat,num,'args_inline'),'arguments + feature YAML':python_config(cat_yaml,num_yaml,'args_feature_yaml'),'main YAML + feature YAML':RegressionTaskConfig.from_yaml(TEST/'configs/regression_one_trial.yaml'),'main YAML + inline features':RegressionTaskConfig.from_yaml(TEST/'configs/regression_inline_features.yaml')}
baseline=next(iter(configs.values())); fields=('engine','model_scope','categorical_columns','numerical_columns','hidden_state_columns','hyperopt','n_trials','metric')
assert all(all(getattr(config,field)==getattr(baseline,field) for field in fields) for config in configs.values())
results={name:RegressionTask(config).train(ROOT/'data/regression/train',ROOT/'data/regression/valid').validation_metrics for name,config in configs.items()}
results
```

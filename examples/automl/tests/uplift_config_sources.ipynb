# Uplift Python and YAML configuration
Inline feature lists and separate absolute YAML feature files resolve to the same typed uplift contract.

```python
from pathlib import Path
import os
from avatar.automl import UpliftTask, UpliftTaskConfig
ROOT=Path.cwd().parent.resolve(); TEST=Path.cwd()/'examples/automl/tests'; os.environ['SBER_INFRA_ROOT']=str(ROOT); data=ROOT/'data/uplift'
cat=[f'cat_feature_{i}' for i in range(1,6)]; num=[f'num_feature_{i}' for i in range(1,6)]
cat_yaml=(TEST/'configs/features/categorical_columns.yaml').resolve(); num_yaml=(TEST/'configs/features/numerical_columns.yaml').resolve(); space={'n_estimators':[6],'max_depth':[2]}
def python_config(categorical_columns,numerical_columns,name):
 return UpliftTaskConfig(target_column='target', client_id_column='epk_id', group_column='group', treatment_column='treatment', inverse_treatment=True, report_month_column='report_month', hidden_state_columns=(), model_scope='product', environment={}, estimate_propensity=False, env_type='local',backend='boosting',engine='xgboost',device='cpu',categorical_columns=categorical_columns,numerical_columns=numerical_columns,hyperopt=True,n_trials=1,search_space=space,verbose=False,output_dir=ROOT/f'outputs/notebook_tests/uplift/{name}')
configs={'arguments + inline features':python_config(cat,num,'args_inline'),'arguments + feature YAML':python_config(cat_yaml,num_yaml,'args_feature_yaml'),'main YAML + feature YAML':UpliftTaskConfig.from_yaml(TEST/'configs/uplift_one_trial.yaml'),'main YAML + inline features':UpliftTaskConfig.from_yaml(TEST/'configs/uplift_inline_features.yaml')}
baseline=next(iter(configs.values())); fields=('engine','model_scope','categorical_columns','numerical_columns','hyperopt','n_trials','metric')
assert all(all(getattr(config,field)==getattr(baseline,field) for field in fields) for config in configs.values())
results={name:UpliftTask(config).train(data/'train',data/'valid').validation_metrics for name,config in configs.items()}
results
```

# Uplift CatBoost and XGBoost
Both engines fit the complete S/T/X composite with one Optuna trial and save native artifacts.

```python
from pathlib import Path
from avatar.automl import UpliftTask,UpliftTaskConfig
ROOT=Path.cwd().parent.resolve(); data=ROOT/'data/uplift'; cat=[f'cat_feature_{i}' for i in range(1,6)]; num=[f'num_feature_{i}' for i in range(1,6)]
spaces={'catboost':{'iterations':[6],'depth':[2]},'xgboost':{'n_estimators':[6],'max_depth':[2]}}; rows=[]
for engine in ('catboost','xgboost'):
 c=UpliftTaskConfig(target_column='target', client_id_column='epk_id', group_column='group', treatment_column='treatment', inverse_treatment=True, report_month_column='report_month', hidden_state_columns=(), model_scope='product', environment={}, env_type='local',backend='boosting',engine=engine,device='cpu',categorical_columns=cat,numerical_columns=num,estimate_propensity=(engine=='xgboost'),hyperopt=True,n_trials=1,search_space=spaces[engine],verbose=False,output_dir=ROOT/f'outputs/notebook_tests/uplift/engines/{engine}')
 t=UpliftTask(c); training=t.train(data/'train',data/'valid'); prediction=t.predict(data/'test'); evaluation=t.evaluate(data/'test',prediction); artifact=t.save(); rows.append((engine,training.validation_metrics,evaluation.metrics,artifact.exists()))
rows
```

# Uplift product, channel and both scopes
Each model part owns a complete S/T/X composite. Product evaluation remains grouped by month and channel; calibration is exercised for product scope.

```python
from pathlib import Path
from fmlib.automl import UpliftTask,UpliftTaskConfig
ROOT=Path.cwd().parent.resolve(); data=ROOT/'data/uplift'; num=[f'num_feature_{i}' for i in range(1,6)]; rows=[]; expected_model_counts={'product':3,'channel':6,'both':9}
for scope in ('product','channel','both'):
 c=UpliftTaskConfig(target_column='target', client_id_column='epk_id', group_column='group', treatment_column='treatment', inverse_treatment=True, report_month_column='report_month', categorical_columns=(), hidden_state_columns=(), environment={}, env_type='local',backend='boosting',engine='xgboost',device='cpu',numerical_columns=num,model_scope=scope,estimate_propensity=False,hyperopt=True,n_trials=1,search_space={'n_estimators':[6],'max_depth':[2]},verbose=False,output_dir=ROOT/f'outputs/notebook_tests/uplift/scopes/{scope}')
 t=UpliftTask(c); training=t.train(data/'train',data/'valid'); assert len(training.validation_metrics)==expected_model_counts[scope]; prediction=t.predict(data/'test'); evaluation=t.evaluate(data/'test',prediction); artifact=t.save(); rows.append((scope,len(training.validation_metrics),prediction.raw_scores is not None,evaluation.metrics_by_group.shape,artifact.exists()))
calibrated_config=UpliftTaskConfig(target_column='target', client_id_column='epk_id', group_column='group', treatment_column='treatment', inverse_treatment=True, report_month_column='report_month', categorical_columns=(), hidden_state_columns=(), environment={}, env_type='local',backend='boosting',engine='xgboost',device='cpu',numerical_columns=num,model_scope='both',estimate_propensity=False,hyperopt=True,n_trials=1,search_space={'n_estimators':[6],'max_depth':[2]},verbose=False,calibration_windows={'2024-10-01':['2024-07-01','2024-08-01','2024-09-01'],'2024-11-01':['2024-08-01','2024-09-01'],'2024-12-01':['2024-09-01']},output_dir=ROOT/'outputs/notebook_tests/uplift/scopes/both_calibrated')
calibrated=UpliftTask(calibrated_config); training=calibrated.train(data/'train',data/'valid'); prediction=calibrated.predict(data/'test',calibration_paths=[data/'train',data/'valid']); evaluation=calibrated.evaluate(data/'test',prediction); artifact=calibrated.save(); assert len(training.validation_metrics)==9 and prediction.raw_scores is not None and evaluation.calibrated_metrics is not None and 'metrics_calibrated' in evaluation.excel_paths; rows.append(('both_calibrated',len(training.validation_metrics),True,evaluation.calibrated_metrics_by_group.shape,artifact.exists()))
rows
```

```python
import os
import pandas as pd
from sklearn.metrics import roc_auc_score

predicts_folder = "predicts"
for predict in os.listdir(predicts_folder):
    df = pd.read_parquet(f"{predicts_folder}/{predict}")
    print("Exp name: ", predict, "RocAucScore", roc_auc_score(df["targets"], df["predicted"]))
```
